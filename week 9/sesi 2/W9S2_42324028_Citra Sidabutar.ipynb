## Penjelasan
pada code program pertama dilakukan instalasi beberapa library yang diperlukan agar program dapat berjalan, yaitu TensorFlow, Matplotlib, dan NumPy. TensorFlow digunakan untuk membuat, mengajar, dan menguji model jaringan saraf. Matplotlib dipakai untuk menampilkan gambar dari dataset dan membuat grafik hasil pelatihan, sedangkan NumPy digunakan untuk membantu dalam mengolah data berbentuk array. Pelaksanaan instalasi diuraikan melalui perintah pip install guna memastikan seluruh pustaka pendukung dapat diintegrasikan secara fungsional ke dalam program.

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import cifar10

# Memuat dataset CIFAR-10
(train_images, train_labels), (test_images, test_labels) = cifar10.load_data()

# Nama kategori pada dataset CIFAR-10
class_names = ['Pesawat', 'Mobil', 'Burung', 'Kucing', 'Rusa',
               'Anjing', 'Katak', 'Kuda', 'Kapal', 'Truk']

# Mengubah label menjadi array 1 dimensi
train_labels = train_labels.flatten()
test_labels = test_labels.flatten()

print("Jumlah data training:", train_images.shape[0])
print("Jumlah data testing:", test_images.shape[0])
print("Ukuran gambar:", train_images.shape[1:])

# Normalisasi nilai piksel dari rentang 0-255 menjadi 0-1
train_images = train_images.astype('float32') / 255.0
test_images = test_images.astype('float32') / 255.0
```

## Penjelasan
pada kode diatas ini menjelaskan Langkah selanjutnya, dimana melibatkan pemanggilan pustaka yang telah terinstal sebelumnya, dilanjutkan dengan proses pemuatan dataset CIFAR-10 melalui fungsi cifar10.load_data(). Koleksi data tersebut dipartisi menjadi 50.000 gambar yang dialokasikan untuk proses pelatihan serta 10.000 gambar yang diperuntukkan bagi keperluan pengujian. Selanjutnya, disusun sebuah daftar yang memuat sepuluh kategori gambar, yakni pesawat terbang, mobil, burung, kucing, rusa, anjing, katak, kuda, kapal, serta truk. Setiap label imej kemudiannya diolah dan ditukarkan ke dalam satu array dimensi tunggal melalui penerapan fungsi flatten(). Selanjutnya, proses normalisasi dilakukan terhadap nilai-nilai piksel pada gambar yang semula berada dalam rentang 0–255, diubah menjadi skala 0–1 melalui pembagian nilai piksel tersebut dengan angka 255. Tujuannya adalah agar data memiliki rentang nilai yang lebih sederhana, sehingga lebih mudah diproses oleh model Neural Network.

```python
plt.figure(figsize=(10, 6))

for i in range(15):
    plt.subplot(3, 5, i + 1)
    plt.imshow(train_images[i])
    plt.title(class_names[train_labels[i]])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

## Penjelasn
Pada sel kode ketiga dilakukan proses visualisasi untuk menampilkan 15 contoh gambar dari dataset CIFAR-10 beserta nama kategorinya. Program menggunakan perulangan for untuk mengambil gambar dari data latihan, lalu menampilkan gambar tersebut dengan menggunakan fungsi imshow(). Setiap gambar diberi judul berdasarkan kategorinya dengan menggunakan perintah plt.title(). Tampilan gambar disusun dalam 3 baris dan 5 kolom agar lebih rapi dan lebih mudah dilihat. Dengan proses ini, saya bisa mengetahui jenis gambar yang ada di dataset dan melihat contoh objek yang akan dienali serta diklasifikasikan oleh model.

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Flatten, Dense, Dropout

# Membuat model Neural Network
model = Sequential([
    Flatten(input_shape=(32, 32, 3)),
    Dense(256, activation='relu'),
    Dropout(0.2),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model.summary()
```

## Penjelasn
Code program ini dibuat model Neural Network dengan beberapa lapisan atau layer yang disusun secara berurutan menggunakan fitur Sequential. Lapisan pertama adalah Flatten yang berperan untuk mengubah gambar dengan ukuran 32 × 32 × 3 menjadi bentuk satu dimensi sehingga bisa diproses oleh model. Selanjutnya, ada Dense dengan 256 neuron dan aktivasi ReLU sebagai lapisan tersembunyi pertama. Setelah itu, Dropout dengan tingkat 0,2 digunakan untuk membantu mengurangi overfitting dengan cara menonaktifkan sebagian neuron secara acak saat proses pelatihan berlangsung. Setelah itu, ada lapisan Dense dengan 128 neuron dan aktivasi ReLU sebagai lapisan tersembunyi yang kedua. Pada bagian terakhir, digunakan Dense dengan 10 neuron dan aktivasi Softmax sebagai lapisan output karena dataset CIFAR-10 memiliki 10 jenis gambar. Perintah model.summary() digunakan untuk menampilkan ringkasan struktur model yang telah dibuat.

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history = model.fit(
    train_images,
    train_labels,
    epochs=10,
    batch_size=64,
    validation_data=(test_images, test_labels)
)
```

## Penjelasan
Pada kode ini dilakukan proses menggabungkan dan melatih model jaringan saraf. Proses penyusunan menggunakan optimizer Adam yang bertugas memperbarui bobot model selama proses belajar, sedangkan sparse_categorical_crossentropy digunakan untuk mengukur tingkat kesalahan prediksi karena data labelnya dalam bentuk angka. Metrik akurasi digunakan untuk mengetahui seberapa tepat model dalam melakukan klasifikasi. Setelah kompilasi selesai, model dilatih dengan menggunakan fungsi model.fit() berdasarkan 50.000 gambar data pelatihan. Pelatihan dilakukan selama 10 siklus dengan ukuran batch 64, yang berarti data diproses secara kelompok-kelompok dengan masing-masing berisi 64 gambar. Pada tahap ini, data validasi juga digunakan untuk mengamati kemajuan kemampuan model sepanjang proses pelatihan. Hasil pelatihan disimpan dalam variabel history dan akan digunakan pada tahap visualisasi.

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)

print(f'Test loss: {test_loss:.4f}')
print(f'Test accuracy: {test_acc:.4f}')
```

## Penjelasan
Pada kode ini, model yang sudah dilatih dilakukan evaluasi menggunakan 10.000 gambar dari data testing. Evaluasi dilakukan dengan menggunakan fungsi model.evaluate() agar mengetahui sejauh mana kemampuan model dalam mengklasifikasikan gambar yang diberikan. Evaluasi tersebut memberikan dua angka, yaitu loss dari tes dan akurasi dari tes. Test loss adalah ukuran seberapa besar kesalahan yang dilakukan model saat memprediksi, sedangkan test accuracy adalah persentase seberapa tepat model dalam mengenali gambar-gambar. Kedua nilai tersebut kemudian ditampilkan dengan perintah print() dan memiliki empat angka di belakang koma. Dalam tahap ini, bisa dilihat seberapa baik model mampu mengenali dan mengklasifikasikan gambar-gambar dari dataset CIFAR-10.

```python
plt.figure(figsize=(12, 4))

# Grafik accuracy
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

# Grafik loss
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

## Penjelasan
Pada kode diatas ini, dilakukan visualisasi hasil pelatihan model dengan menggunakan dua grafik, yaitu grafik akurasi dan grafik loss. Grafik akurasi digunakan untuk menunjukkan bagaimana tingkat keakuratan model berubah selama proses pelatihan dan validasi selama 10 epoch. Grafik loss digunakan untuk memantau perubahan tingkat kesalahan model selama proses pembelajaran. Data grafik diambil dari variabel history yang menampung hasil pelatihan dari sel sebelumnya. Kedua grafik ditampilkan berdampingan agar lebih mudah dibandingkan. Dengan bantuan visualisasi ini, saya bisa melihat bagaimana kemampuan model berkembang selama proses pelatihan dan mengetahui apakah ada tanda-tanda overfitting, yaitu saat model terlalu baik menghafal data latihan tetapi kurang mampu mengenali data baru yang belum pernah dilihat sebelumnya.

```python
predictions = model.predict(test_images[:25])

plt.figure(figsize=(12, 12))

for i in range(25):
    predicted_label = np.argmax(predictions[i])
    true_label = test_labels[i]

    plt.subplot(5, 5, i + 1)
    plt.imshow(test_images[i])
    plt.title(
        f'Prediksi: {class_names[predicted_label]}\n'
        f'Asli: {class_names[true_label]}',
        fontsize=8,
        color=('green' if predicted_label == true_label else 'red')
    )
    plt.axis('off')

plt.tight_layout()
plt.show()
```

## Penjelasan
Pada kode program terakhir ini, dilakukan prediksi terhadap 25 gambar pertama dari data uji menggunakan fungsi model.predict(). Model akan mengeluarkan angka kemungkinan untuk setiap kelompok, lalu fungsi np.argmax() digunakan untuk memilih kelompok yang memiliki angka kemungkinan paling besar sebagai hasilnya. Hasil tersebut kemudian dibandingkan dengan label sebenarnya yang ada di dalam dataset. Program menampilkan gambar beserta nama kategori hasil prediksi dan kategori aslinya. Ketika hasil prediksi selaras dengan label aktual, representasi teks akan diwarnai hijau; sebaliknya, apabila terdapat ketidaksesuaian, tampilannya akan disorot dengan warna merah. Pada fase ini, saya dapat mengamati secara langsung bagaimana model menunjukkan kemampuannya dalam mengidentifikasi serta mengklasifikasikan gambar ke dalam kategori yang relevan.

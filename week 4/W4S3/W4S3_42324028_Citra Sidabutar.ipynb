```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Load dataset Iris dari scikit-learn
iris = load_iris()
```

# Penjelasan
Kode ini digunakan untuk mengimpor library yang diperlukan dalam praktikum. numpy digunakan untuk mengolah data berupa angka, pandas digunakan untuk mengelola data dalam bentuk tabel, sedangkan load_iris digunakan untuk mengambil dataset Iris dari scikit-learn. Setelah itu, dataset Iris dimasukkan ke dalam variabel iris sehingga dapat digunakan pada proses selanjutnya.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

# Penjelasan
Pada kode ini, data dari dataset Iris diubah terlebih dahulu menjadi bentuk DataFrame menggunakan pd.DataFrame(). Data fitur dari iris.data dimasukkan ke dalam df, sedangkan iris.feature_names digunakan sebagai nama kolomnya. Setelah itu, kolom Species ditambahkan untuk menyimpan nama jenis bunga berdasarkan target yang terdapat pada dataset.
Selanjutnya, X digunakan untuk menyimpan data fitur dengan menghapus kolom Species, sehingga yang tersisa adalah ukuran sepal dan petal. Sementara itu, y digunakan untuk menyimpan kolom Species sebagai target atau label. Terakhir, print(X) digunakan untuk menampilkan seluruh data fitur yang sudah dipisahkan dari label.

```python
y
```

# Penjelasan
y merupakan variabel yang digunakan untuk menyimpan target atau label dari dataset. Pada kode sebelumnya, y berisi nama species dari bunga Iris, yaitu setosa, versicolor, dan virginica. Setelah menggunakan LabelEncoder, isi y diubah menjadi angka 0, 1, dan 2 agar lebih mudah diproses oleh model Naïve Bayes.

```python
from sklearn.preprocessing import LabelEncoder

lab = LabelEncoder()
y = lab.fit_transform(y)
print(y)
```

# Penjelasan
Pada bagian ini, kode digunakan untuk mengubah nama species yang masih berbentuk teks menjadi angka. LabelEncoder() membuat alat yang digunakan untuk melakukan perubahan tersebut, kemudian fit_transform(y) membaca semua label yang ada pada y dan memberikan kode angka untuk setiap jenis species. Jadi, setosa, versicolor, dan virginica akan diubah menjadi 0, 1, dan 2. Hasil perubahan tersebut disimpan kembali ke y, sehingga data label sudah berbentuk angka dan bisa digunakan dalam proses pembuatan model Naïve Bayes.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

# Penjelasan
kode diatas  ini digunakan untuk membagi data menjadi data training dan testing. test_size=0.2 berarti 20% data digunakan sebagai data testing, sedangkan 80% sisanya digunakan untuk training. random_state=0 digunakan agar pembagian data tetap sama setiap kali kode dijalankan. Pembagian dilakukan untuk X sebagai fitur dan y sebagai label atau target.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

# Penjelasan
Pada kode ini, GaussianNB code disini digunakan untuk membuat model klasifikasi dengan algoritma Gaussian Naïve Bayes. Setelah model dibuat, model.fit(X_train, y_train) akan digunakan untuk melatih model menggunakan data training, yaitu X_train sebagai data fitur dan y_train sebagai label atau jenis species. Hasil model yang sudah dilatih kemudian disimpan ke dalam variabel NB_model, sehingga model tersebut dapat digunakan untuk melakukan prediksi pada data testing maupun data baru.

```python
result = NB_model.predict(X_test)
print(result)
```

# Penjelasan 
Pada kode ini, NB_model.predict(X_test) digunakan untuk melakukan prediksi terhadap data testing yang sebelumnya sudah dipisahkan dari dataset. Hasil prediksi tersebut disimpan ke dalam variabel result. Selanjutnya, print(result) digunakan untuk menampilkan hasil prediksi dalam bentuk angka, yaitu 0, 1, atau 2, yang masing-masing mewakili jenis species pada dataset Iris.

```python
from sklearn.metrics import accuracy_score

print('Accuracy of data testing: ', accuracy_score(y_test, result) * 100)
```

# Penjelasan
Pada kode ini, accuracy_score code digunakan untuk menghitung tingkat ketepatan hasil prediksi model terhadap data sebenarnya pada y_test. Hasil prediksi yang terdapat pada result dibandingkan dengan data sebenarnya, kemudian hasilnya dikalikan 100 agar ditampilkan dalam bentuk persentase. Dari hasil yang terdapat pada modul, akurasi model adalah 96,67%, yang menunjukkan bahwa sebagian besar data testing berhasil diprediksi dengan benar.

```python
# Prediksi inputan data baru
# input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Pada kode ini, data baru [5.1, 3.5, 1.4, 0.2] dimasukkan ke dalam variabel input_data dalam bentuk list. Data tersebut kemudian digunakan sebagai input untuk melakukan prediksi menggunakan model NB_model yang sebelumnya sudah dilatih dengan dataset Iris. Setelah proses prediksi dilakukan, hasilnya masih berupa angka sehingga perlu diubah kembali menjadi nama species menggunakan lab.inverse_transform(). Dari hasil prediksi, data tersebut termasuk ke dalam species Iris setosa.

```python
# Menggunakan DataFrame agar tidak memicu peringatan (UserWarning)
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

# Penjelasan
Pada kode ini, data baru [5.1, 3.5, 1.4, 0.2] dibuat dalam bentuk DataFrame dengan nama kolom yang sama seperti dataset Iris agar tidak muncul warning. Setelah itu, NB_model.predict() digunakan untuk memprediksi species dari data tersebut. Hasil prediksi yang masih berupa angka kemudian diubah menjadi nama species menggunakan lab.inverse_transform(). Hasil akhirnya adalah Iris setosa.
